```python
def evaluar_perfil(edad, le_gusta_musica, le_gusta_bailar):
    # Condición 1: 18 o más años, le gusta la música Y le gusta bailar
    if edad >= 18 and le_gusta_musica and le_gusta_bailar:
        return "Eres una estrella de la fiesta"
    # Condición 2: Le gusta la música, pero NO le gusta bailar (independiente de la edad)
    elif le_gusta_musica and not le_gusta_bailar:
        return "Te gusta la música, pero no tanto bailar"
    # Cualquier otro caso
    else:
        return "Tienes tu propio estilo"


# --- Entrada de datos por parte del usuario ---
edad = int(input("Ingresa tu edad: "))

# Preguntamos y convertimos las respuestas a valor booleano (True/False)
respuesta_musica = input("¿Te gusta la música? (si/no): ").strip().lower()
le_gusta_musica = respuesta_musica == "si"

respuesta_bailar = input("¿Te gusta bailar? (si/no): ").strip().lower()
le_gusta_bailar = respuesta_bailar == "si"

# Llamada a la función e impresión del resultado
mensaje = evaluar_perfil(edad, le_gusta_musica, le_gusta_bailar)
print(f"\nResultado: {mensaje}")
```
