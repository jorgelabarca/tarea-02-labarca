```python
# Recorrer los números del 1 al 100
for numero in range(1, 101):
    # Evaluar primero si es divisible por 10 (lo que implica que también es divisible por 5)
    if numero % 10 == 0:
        print(f"{numero} yep yep")
    # Si no es divisible por 10 pero sí por 5
    elif numero % 5 == 0:
        print(f"{numero} yep")
```
